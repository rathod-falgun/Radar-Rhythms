```python
import os
import csv
import json
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

OUTPUT_DIR = os.path.join("..", "data", "output")
CSV_PATH = os.path.join(OUTPUT_DIR, "nisar_flood_timeseries.csv")
JSON_PATH = os.path.join(OUTPUT_DIR, "nisar_flood_timeseries_report.json")

df = pd.read_csv(CSV_PATH)
df
```

## 1. Multi-Temporal Flood Hydrograph
Comparing the NISAR radar-derived inundation area ($km^2$) against hydrological river discharge ($m^3/s$).

```python
hydro_img = Image.open(os.path.join(OUTPUT_DIR, "nisar_flood_hydrograph.png"))
plt.figure(figsize=(14, 10), dpi=150)
plt.imshow(hydro_img)
plt.axis("off")
plt.title("NISAR Multi-Temporal Flood Hydrograph (June - August 2026)", fontsize=14, fontweight="bold")
plt.show()
```

## 2. 7-Cycle Inundation Taxonomy Progression Mosaic
Visualizing the spatial expansion and contraction of floodwaters across Kaziranga corridor across all 7 revisit cycles.

```python
mosaic_img = Image.open(os.path.join(OUTPUT_DIR, "nisar_temporal_mosaic.png"))
plt.figure(figsize=(16, 9), dpi=150)
plt.imshow(mosaic_img)
plt.axis("off")
plt.title("7-Revisit Inundation Classification Progression", fontsize=14, fontweight="bold")
plt.show()
```
